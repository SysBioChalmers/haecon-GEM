# Relabel unsupported reactions

Mark 93 retinoid/vision reactions as 'Unsupported' in all 9 contextualized models.
These reactions lack H. contortus support and should be excluded from downstream analyses.

```python
import cobra

MODEL_DIR = "model/contextualized_models"
MODELS = [
    "egg_c.xml", "L1_c.xml", "L2_c.xml", "L3_c.xml",
    "L4f_c.xml", "L4m_c.xml", "Am_c.xml", "Af_c.xml", "L4mf_union_c.xml"
]

print(f"Models directory: {MODEL_DIR}")
print(f"Total models to process: {len(MODELS)}")
```

```python
# 93 unsupported retinoid/vision reactions
UNSUPPORTED = {
    "MAR00023", "MAR01506", "MAR01507", "MAR02960", "MAR02993", "MAR03031",
    "MAR03036", "MAR03041", "MAR03131", "MAR06630", "MAR06631", "MAR06632",
    "MAR06633", "MAR06634", "MAR06635", "MAR06636", "MAR06637", "MAR06638",
    "MAR06639", "MAR06641", "MAR06642", "MAR06646", "MAR06647", "MAR06648",
    "MAR06649", "MAR06651", "MAR06652", "MAR06653", "MAR06654", "MAR06655",
    "MAR06656", "MAR06657", "MAR06658", "MAR06659", "MAR06660", "MAR06662",
    "MAR06663", "MAR06664", "MAR06665", "MAR06666", "MAR06667", "MAR06668",
    "MAR06669", "MAR06670", "MAR06671", "MAR06672", "MAR06673", "MAR06674",
    "MAR06675", "MAR06676", "MAR06677", "MAR06678", "MAR06680", "MAR06682",
    "MAR06685", "MAR06686", "MAR06687", "MAR06690", "MAR06691", "MAR06692",
    "MAR06693", "MAR06694", "MAR06695", "MAR06697", "MAR06699", "MAR06700",
    "MAR06701", "MAR06702", "MAR06703", "MAR06704", "MAR06705", "MAR08697",
    "MAR08698", "MAR08699", "MAR08702", "MAR08703", "MAR08704", "MAR08706",
    "MAR08709", "MAR08710", "MAR08711", "MAR08712", "MAR08716", "MAR11894",
    "MAR11895", "MAR11896", "MAR11897", "MAR20001", "MAR20003", "MAR20005",
    "MAR20006", "MAR20007", "MAR20008",
}

print(f"Unsupported reactions to relabel: {len(UNSUPPORTED)}")
```

```python
# Relabel in each model
for model_file in MODELS:
    model = cobra.io.read_sbml_model(f"{MODEL_DIR}/{model_file}")
    relabeled = 0
    
    for rxn_id in UNSUPPORTED:
        if rxn_id in model.reactions:
            model.reactions.get_by_id(rxn_id).subsystem = "Unsupported"
            relabeled += 1
    
    cobra.io.write_sbml_model(model, f"{MODEL_DIR}/{model_file}")
    print(f"{model_file}: {relabeled} reactions relabeled")
```
